```python
def main():
    value = 99
    print(f'The value is {value}')
    change_me(value)

def change_me(arg):
    print(f'value of arg is {arg}')
    print('I am changing value')
    arg = 0
    print(f'Now the value is {arg}')

#call the main function    
main()
```
