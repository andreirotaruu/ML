```python
import pandas as pd

df = pd.read_csv('iris.csv', names=['sepal length','sepal width','petal length','petal width','target'], header=1)
df.head()
```

```python
# X is data frame of features; y is data frame of target
features = ['sepal length','sepal width','petal length','petal width']
X = df[features]
y = df["target"]
```

```python
y.value_counts()
```

```python
X[0:5]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)
```

```python
from sklearn import tree
from sklearn.naive_bayes import GaussianNB
model = tree.DecisionTreeClassifier()
#model = GaussianNB()

model
```

```python
model.fit(X_train,y_train)
```

```python
y_predict = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
print(f"Accuracy score: {accuracy_score(y_test,y_predict)}")
# for binary classification use below :
#print(f"precision score: {precision_score(y_test,y_predict)}")
#print(f"Recall score: {recall_score(y_test,y_predict)}")
#print(f"F1 score: {f1_score(y_test,y_predict)}")
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test,y_predict))
```

```python
conf_matrix = confusion_matrix(y_test,y_predict)
cm_df = pd.DataFrame(conf_matrix)

cm_df
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

ConfusionMatrixDisplay.from_predictions(y_test,y_predict)
plt.show()
```

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(model,X,y,scoring="accuracy", cv=5)
scores
```
