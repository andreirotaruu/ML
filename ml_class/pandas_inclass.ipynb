# Import the necessary libraries

```python
import pandas as pd
```

# Load the Data

```python
employee_df = pd.read_csv('EMPLOYEE.csv')
department_df = pd.read_csv('DEPARTMENT.csv')
```

# Data Cleaning 

```python
#check for missing values 
employee_df.isnull().sum()
department_df.isnull().sum()
```

```python
#drop na values
employee_df.dropna()
department_df.dropna()
```

```python
#ensure all DNO in employee dataframe matches with DNUMBER in department dataframe
employee_df["DNO"].isin(department_df["DNUMBER"]).all()
```

# Data Manipulation

```python
#get employee names with salaries over 40000

salary_row = employee_df[employee_df['SALARY']>40000]
salary_row[['FNAME', 'LNAME']]
```

```python
#get employee names whose supervisor is 666666602

supervisor_row = employee_df[employee_df['SUPERSSN'] == 666666602]
supervisor_row[['FNAME', 'LNAME']]
```

# Merging the data

```python
#merge dfs on DNO and DNUMBER
merge_df = pd.merge(employee_df, department_df, left_on='DNO', right_on='DNUMBER')
merge_df.head()
```

# Filter for the research department

```python
#filter the dataframe
research_df = merge_df[merge_df['DNAME'] == 'Research']

#show employees that are in research department 
print(research_df[['FNAME', 'LNAME']])
```
